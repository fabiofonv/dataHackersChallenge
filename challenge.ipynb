```python
# dict_faixasalarial = {
#     "de R$ 8.001/mês a R$ 12.000/mês": 10000,
#     "de R$ 12.001/mês a R$ 16.000/mês": 14000,
#     "de R$ 6.001/mês a R$ 8.000/mês": 7000,
#     "de R$ 4.001/mês a R$ 6.000/mês": 5000,
#     "de R$ 16.001/mês a R$ 20.000/mês": 18000,
#     "de R$ 3.001/mês a R$ 4.000/mês": 3500,
#     "de R$ 20.001/mês a R$ 25.000/mês": 22500,
#     "de R$ 2.001/mês a R$ 3.000/mês": 2500,
#     "de R$ 25.001/mês a R$ 30.000/mês": 27500,
#     "de R$ 30.001/mês a R$ 40.000/mês": 35000,
#     "Acima de R$ 40.001/mês": 40000,
#     "Menos de R$ 1.000/mês": 1000,
#     "de R$ 1.001/mês a R$ 2.000/mês": 1500,
# }
# df_rem_limpo["salario_medio"] = df_rem_limpo["faixa_salarial"].map(dict_faixasalarial)
# df_rem_limpo["salario_medio"].value_counts()

def extrair_valor_medio_salario(faixa_salarial):
    """Extrai o valor médio da faixa salarial (R$)"""
    if pd.isna(faixa_salarial):
        return np.nan

    mapeamento = {
        "de R$ 8.001/mês a R$ 12.000/mês": 10000,
        "de R$ 12.001/mês a R$ 16.000/mês": 14000,
        "de R$ 6.001/mês a R$ 8.000/mês": 7000,
        "de R$ 4.001/mês a R$ 6.000/mês": 5000,
        "de R$ 16.001/mês a R$ 20.000/mês": 18000,
        "de R$ 3.001/mês a R$ 4.000/mês": 3500,
        "de R$ 20.001/mês a R$ 25.000/mês": 22500,
        "de R$ 2.001/mês a R$ 3.000/mês": 2500,
        "de R$ 25.001/mês a R$ 30.000/mês": 27500,
        "de R$ 30.001/mês a R$ 40.000/mês": 35000,
        "Acima de R$ 40.001/mês": 40000,
        "Menos de R$ 1.000/mês": 1000,
        "de R$ 1.001/mês a R$ 2.000/mês": 1500,
    }
    return mapeamento.get(faixa_salarial, np.nan)

df_rem_limpo["salario_medio"] = df_rem_limpo["faixa_salarial"].apply(extrair_valor_medio_salario)
```

```python
labels_curtos = {
    "de R$ 8.001/mês a R$ 12.000/mês": "8-12K",
    "de R$ 12.001/mês a R$ 16.000/mês": "12-16K",
    "de R$ 6.001/mês a R$ 8.000/mês": "6-8K",
    "de R$ 4.001/mês a R$ 6.000/mês": "4-6K",
    "de R$ 16.001/mês a R$ 20.000/mês": "16-20K",
    "de R$ 3.001/mês a R$ 4.000/mês": "3-4K",
    "de R$ 20.001/mês a R$ 25.000/mês": "20-25K",
    "de R$ 2.001/mês a R$ 3.000/mês": "2-3K",
    "de R$ 25.001/mês a R$ 30.000/mês": "25-30K",
    "de R$ 30.001/mês a R$ 40.000/mês": "30-40K",
    "Acima de R$ 40.001/mês": ">40K",
    "Menos de R$ 1.000/mês": "<1K",
    "de R$ 1.001/mês a R$ 2.000/mês": "1-2K",
}
df_rem_limpo["short_faixa_salarial"] = df_rem_limpo["faixa_salarial"].map(labels_curtos)
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20,8))

counts = df_rem_limpo["faixa_salarial"].value_counts(sort=False)
ax1.bar(range(len(counts)), counts.values, color="skyblue", alpha=0.7, edgecolor="navy", linewidth=1)
ax1.set_title("Distribuição de faixa salarial", fontsize=16, fontweight="bold")
ax1.set_xlabel("Faixa salarial")
ax1.set_ylabel("Quantidade")
ax1.set_xticks(range(len(counts)))
ax1.set_xticklabels([labels_curtos[faixa] for faixa in counts.index], rotation=45)
ax1.grid(True, alpha=0.3)
for i, v in enumerate(counts.values):
    ax1.text(i, v + 10, str(v), ha="center", fontweight="bold")

box_data = df_rem_limpo["salario_medio"].dropna()
ax2.boxplot(box_data, vert=True, patch_artist=True,
            boxprops=dict(facecolor="lightcoral", alpha=0.7),
            medianprops=dict(color="red", linewidth=2))
ax2.set_title("Distribuição Salarial Média", fontsize=16, fontweight="bold")
ax2.set_ylabel("Salário R$")
ax2.grid(True, alpha=0.3)
ax2.text(1.1, box_data.max()*0.8,
         f"Média: R${box_data.mean():.0f}\nMediana: R${box_data.median():.0f}\nDesvio: R${box_data.std():.0f}",
         fontsize=10, bbox=dict(boxstyle="round,pad=0.3" , facecolor="yellow", alpha=0.7))
plt.tight_layout()
plt.show()
```
